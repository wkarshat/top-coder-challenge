# Output Visualization of Predictions
Shows distributions, outliers, and summary of output results.

```python
import pandas as pd
import matplotlib.pyplot as plt
import json

with open('private_cases.json') as f:
    data = json.load(f)
df = pd.DataFrame([r['input'] if 'input' in r else r for r in data])
# Simulate predictions
def dummy_predict(row):
    return row['trip_duration_days'] * 100 + row['miles_traveled'] * 0.5 + row['total_receipts_amount']
df['prediction'] = df.apply(dummy_predict, axis=1)
plt.hist(df['prediction'], bins=30, color='green', alpha=0.7)
plt.title('Prediction Distribution')
plt.xlabel('Predicted Value')
plt.ylabel('Count')
plt.show()

df.boxplot(column='prediction')
plt.title('Prediction Boxplot')
plt.show()
display(df['prediction'].describe())
```
