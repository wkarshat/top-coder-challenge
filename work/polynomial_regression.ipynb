# Polynomial Regression Analysis
This notebook explores a 2nd-degree polynomial regression fit, visualizes residuals, and helps determine whether curvilinear terms help explain the legacy system.

```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score, mean_absolute_error
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

df = pd.read_csv('public.csv')
X = df[['trip_duration_days', 'miles_traveled', 'total_receipts_amount']]
y = df['expected_output']
display(df.head())
```

## Polynomial Features & Fit

```python
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)
reg_poly = LinearRegression().fit(X_poly, y)
df['poly_pred'] = reg_poly.predict(X_poly)
df['poly_resid'] = df['expected_output'] - df['poly_pred']
print('Polynomial R^2 score:', r2_score(y, df['poly_pred']))
print('Mean Absolute Error:', mean_absolute_error(y, df['poly_pred']))
feature_names = poly.get_feature_names_out(X.columns)
coef_table = pd.DataFrame({'feature': feature_names, 'coef': reg_poly.coef_})
display(coef_table.head(10))
```

## Visualization: Predicted vs Actual, Residuals

```python
plt.figure(figsize=(8,6))
plt.scatter(df['expected_output'], df['poly_pred'], s=8, alpha=0.7)
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', label='Perfect Fit')
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.title('Polynomial Regression: Actual vs. Predicted')
plt.legend()
plt.show()

# Residuals
plt.figure(figsize=(8,6))
plt.scatter(df['poly_pred'], df['poly_resid'], s=8, alpha=0.7)
plt.axhline(0, color='red', lw=1)
plt.xlabel('Predicted')
plt.ylabel('Residual (Actual - Predicted)')
plt.title('Polynomial Regression Residuals')
plt.show()
```

## Discussion
- Polynomial regression can capture some curvature in the legacy system, but residual patterns may still reveal missing rules, bonuses, or caps.
- Compare R^2 and error to the linear model.
