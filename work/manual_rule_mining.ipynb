# Manual Rule Mining & Data Exploration
Use visual, descriptive, and filter-based exploration to hypothesize legacy rules.

```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
df = pd.read_csv('public.csv')
display(df.head())
```

## Summarize by Trip Length

```python
summary = df.groupby('trip_duration_days')['expected_output'].agg(['mean', 'std', 'count'])
display(summary)
plt.figure(figsize=(8,6))
plt.errorbar(summary.index, summary['mean'], yerr=summary['std'], fmt='o-')
plt.xlabel('Trip Duration Days')
plt.ylabel('Mean Output (± std)')
plt.title('Output by Trip Duration Days')
plt.show()
```

## 5-Day Bonus Exploration

```python
five_day = df[df['trip_duration_days'] == 5]
display(five_day.sort_values('expected_output').head(20))
```

## Compare for Same Miles, Different Receipts

```python
same_miles = df[df['miles_traveled'] == 100]
display(same_miles[['total_receipts_amount', 'expected_output']].sort_values('total_receipts_amount'))
```

## Visualize Output Distribution

```python
plt.figure(figsize=(8,4))
plt.hist(df['expected_output'], bins=30, color='skyblue')
plt.xlabel('Expected Output')
plt.title('Distribution of Reimbursement Outputs')
plt.show()
```
