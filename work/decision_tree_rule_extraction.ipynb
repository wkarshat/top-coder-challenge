# Decision Tree Regression for Rule Extraction
Reveal if-then-else and bonus/cap patterns in legacy output.

```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor, plot_tree, export_text
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, r2_score

df = pd.read_csv('public.csv')
X = df[['trip_duration_days', 'miles_traveled', 'total_receipts_amount']]
y = df['expected_output']

tree = DecisionTreeRegressor(max_depth=5, random_state=42)
tree.fit(X, y)
df['tree_pred'] = tree.predict(X)
print(export_text(tree, feature_names=list(X.columns)))
print('Tree R^2:', r2_score(y, df['tree_pred']))
print('Mean Absolute Error:', mean_absolute_error(y, df['tree_pred']))
```

## Visualization: Tree Plot, Fit Quality

```python
plt.figure(figsize=(16, 8))
plot_tree(tree, feature_names=X.columns, filled=True, rounded=True, max_depth=3)
plt.title('Decision Tree (First 3 Levels)')
plt.show()

plt.figure(figsize=(8,6))
plt.scatter(df['expected_output'], df['tree_pred'], s=8, alpha=0.6)
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', label='Perfect Fit')
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.title('Decision Tree: Actual vs. Predicted')
plt.legend()
plt.show()
```
