# Piecewise (Segmented) Regression Example
Fit breakpoints in `miles_traveled` to reveal non-linear jumps and possible tiered reimbursement rates.

```python
import pandas as pd
import numpy as np
import pwlf  # pip install pwlf
import matplotlib.pyplot as plt
from IPython.display import display

df = pd.read_csv('public.csv')
x = df['miles_traveled'].values
y = df['expected_output'].values
display(df[['miles_traveled', 'expected_output']].head())
```

## Fit Piecewise Model (3 Segments)

```python
my_pwlf = pwlf.PiecewiseLinFit(x, y)
breaks = my_pwlf.fit(3)
y_hat = my_pwlf.predict(x)
print('Segment breakpoints:', breaks)
print('Segment slopes:', my_pwlf.slopes)
```

## Visualization: Piecewise Fit

```python
plt.figure(figsize=(10,6))
plt.scatter(x, y, s=8, alpha=0.4, label='Actual')
plt.plot(x, y_hat, color='red', lw=2, label='Piecewise Fit')
plt.xlabel('Miles Traveled')
plt.ylabel('Expected Output')
plt.title('Piecewise Linear Regression (Miles vs Output)')
plt.legend()
plt.show()
```

## Discussion
