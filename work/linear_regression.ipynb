# Linear Regression Analysis for Legacy Reimbursement System
This notebook applies linear regression to the legacy reimbursement data.
You will see: data exploration, regression fit, residuals, visualizations, and key findings.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error
from IPython.display import display

# Load Data
df = pd.read_csv('public.csv')
display(df.head())
```

## Data Exploration
Visualize variable distributions and relationships.

```python
# Histograms
df.hist(bins=20, figsize=(12,6))
plt.tight_layout()
plt.show()

# Pairplot for variables
sns.pairplot(df, vars=['trip_duration_days', 'miles_traveled', 'total_receipts_amount', 'expected_output'])
plt.show()
```

## Linear Regression Fit
Model the relationship as a weighted sum of the three variables.

```python
X = df[['trip_duration_days', 'miles_traveled', 'total_receipts_amount']]
y = df['expected_output']

reg = LinearRegression().fit(X, y)
df['linear_pred'] = reg.predict(X)
df['linear_resid'] = df['expected_output'] - df['linear_pred']

# Model summary
print('Coefficients:', dict(zip(X.columns, reg.coef_)))
print('Intercept:', reg.intercept_)
print('R^2 score:', r2_score(y, df['linear_pred']))
print('Mean Absolute Error:', mean_absolute_error(y, df['linear_pred']))
```

## Visualization: Actual vs Predicted, Residuals

```python
# Actual vs Predicted
plt.figure(figsize=(8,6))
plt.scatter(df['expected_output'], df['linear_pred'], s=8, alpha=0.7)
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--', label='Perfect Fit')
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.title('Linear Regression: Actual vs. Predicted')
plt.legend()
plt.show()

# Residuals
plt.figure(figsize=(8,6))
plt.scatter(df['linear_pred'], df['linear_resid'], s=8, alpha=0.7)
plt.axhline(0, color='red', lw=1)
plt.xlabel('Predicted')
plt.ylabel('Residual (Actual - Predicted)')
plt.title('Linear Regression Residuals')
plt.show()
```

## Findings
- Linear regression provides a baseline fit and highlights which effects are not captured by a simple linear model.
- Check residuals for systematic bias (bonus zones, caps, etc.).
